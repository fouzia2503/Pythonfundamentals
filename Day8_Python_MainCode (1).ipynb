```python
useraccount={'username':'codetraining','firstname':'code','lastname':'training','dob':'01-01-2020','pwd':'54321'}
```

```python
print(useraccount)
```

```python
for a,b in useraccount.items():
    print(a)
    print(b)
```

```python
for a,b in useraccount.items():
    print(a,b)
```

```python
for a,b in useraccount.items():
    print(f"Key:{a}")
    print(f"value:{b}")
```

```python
for a,b in useraccount.items():
    print(f"Key:{a}\t,value:{b}\t")
   
```

```python
for a,b in useraccount.items():
    print(f"Key:{a}")
    print(f"value:{b}\n")
```

```python
#I want to get only the keys:
```

```python
for a in useraccount.keys():
    print(a)
```

```python
for a in useraccount.keys():
    print(f"key:{a}")
```

```python
for b in useraccount.values():
    print(b)
```

```python
for b in useraccount.values():
    print(f"value:{b}")
```
